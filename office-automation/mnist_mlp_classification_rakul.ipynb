MNIST Handwritten Digit Classification using MLP


### Cell 1: Import Libs

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt

print("TensorFlow Version:", tf.__version__)
```

### Cell 2: Load and Preprocess Dataset

```python
# Load MNIST raw digits data
mnist = tf.keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = mnist.load_data()

# Normalize pixel channels from range [0, 255] to [0.0, 1.0]
X_train = X_train / 255.0
X_test = X_test / 255.0

print("--- Data Dimensions ---")
print(f"Training shapes: Features {X_train.shape}, Labels {y_train.shape}")
print(f"Testing shapes:  Features {X_test.shape}, Labels {y_test.shape}")
```

### Cell 3: Build the Network Architecture

```python
# Construct structural layer stack
model = models.Sequential([
    # Flatten 28x28 image grid matrices to a 784-element 1D array
    layers.Flatten(input_shape=(28, 28)),
    
    # First dense layer mapping features using Rectified Linear Units (ReLU)
    layers.Dense(128, activation='relu'),
    
    # Second dense layer narrowing down network features
    layers.Dense(64, activation='relu'),
    
    # Softmax output mapping probability distributions across 10 classes
    layers.Dense(10, activation='softmax')
])

# Setup optimization metrics and loss constraints
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Print structural layers summary
model.summary()
```

### Cell 4: Train and Evaluate

```python
# Run model training loop over 5 execution iterations
history = model.fit(
    X_train, 
    y_train, 
    epochs=5, 
    validation_split=0.1, 
    batch_size=64
)

# Benchmark prediction precision against unseen testing splits
test_loss, test_acc = model.evaluate(X_test, y_test, verbose=0)
print(f"\nFinal Verified Testing Accuracy Score: {test_acc:.4f}")
```
