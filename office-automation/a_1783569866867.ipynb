```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler, StandardScaler, OneHotEncoder, LabelEncoder
```

```python
df= pd.read_csv("HousingData.csv")
df.sample(10)
print(df.isnull().sum())
df['CRIM'] = df['CRIM'].fillna(df['CRIM'].median())
df['ZN'] = df['ZN'].fillna(df['ZN'].median())
df['INDUS'] = df['INDUS'].fillna(df['INDUS'].median())
df['CHAS'] = df['CHAS'].fillna(df['CHAS'].median())
df['AGE'] = df['AGE'].fillna(df['AGE'].median())
df['LSTAT'] = df['LSTAT'].fillna(df['LSTAT'].median())
print(df.isnull().sum())
df.sample(10)
```

```python
Q1 = df['MEDV'].quantile(0.25)
Q3 = df['MEDV'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
df['MEDV'] = np.where(df['MEDV'] < lower, lower, np.where(df['MEDV'] > upper, upper, df['MEDV']))
print(df['MEDV'].describe())
```

```python
df1 = pd.read_csv("Titanic-Dataset.csv")
df1['Age'] = df1['Age'].fillna(df1['Age'].median())

minmax_scaler = MinMaxScaler()
df1['Age_MinMax'] = minmax_scaler.fit_transform(df1[['Age']])

std_scaler = StandardScaler()
df1['Age_Standard'] = std_scaler.fit_transform(df1[['Age']])

print(df1[['Age', 'Age_MinMax', 'Age_Standard']].head())
```

```python
iris = pd.read_csv("csv_data_iris.csv")

l= LabelEncoder()
iris['cat'] = l.fit_transform(iris['class'])

ohe = OneHotEncoder(sparse_output=False)
ohe_array = ohe.fit_transform(iris[['class']])
print(ohe_array)
iris.sample(10)
```

```python
print(df1['Survived'].value_counts())

from sklearn.utils import resample

died = df1[df1['Survived'] == 0]
survived = df1[df1['Survived'] == 1]
survived_upsampled = resample(survived, replace=True, n_samples=len(died), random_state=42)
df1_balanced = pd.concat([died, survived_upsampled])


print(df1_balanced['Survived'].value_counts())
```
