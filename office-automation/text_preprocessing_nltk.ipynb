```python
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
```

```python
text = """
Cristiano Ronaldo is a famous Portuguese football player. He was born in Madeira, Portugal, on February 5, 1985. Ronaldo started his professional career with Sporting CP before moving to Manchester United. He later played for Real Madrid and Juventus, and he currently plays for Al Nassr. Ronaldo is known for his speed, goal scoring ability, powerful shooting, and dedication to football. He has won many trophies and individual awards throughout his career. He has also represented Portugal in several major international tournaments and helped Portugal win the UEFA European Championship in 2016. His career is widely recognized for remarkable goal-scoring records and longevity.

"""

print(text)
```

```python
text_lower = text.lower()

print(text_lower)
```

```python
tokens = word_tokenize(text_lower)

print("Tokens:")
print(tokens)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word for word in tokens
    if word.isalpha() and word not in stop_words
]

print("After Stop-word Removal:")
print(filtered_tokens)
```

```python
stemmer = PorterStemmer()

stemmed_words = [
    stemmer.stem(word)
    for word in filtered_tokens
]

print("After Stemming:")
print(stemmed_words)
```

```python
preprocessed_text = " ".join(stemmed_words)

print("Final Preprocessed Text:")
print(preprocessed_text)
```

```python
print("Original Text:")
print(text)

print("\nTokenized Text:")
print(tokens)

print("\nAfter Stop-word Removal:")
print(filtered_tokens)

print("\nAfter Stemming:")
print(stemmed_words)
```
