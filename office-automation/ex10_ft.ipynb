```python
import torch

from transformers import (
    T5Tokenizer,
    T5ForConditionalGeneration
)
```

```python
model_path = "./t5_summarizer"

tokenizer = T5Tokenizer.from_pretrained(
    model_path
)

model = T5ForConditionalGeneration.from_pretrained(
    model_path
)
```

```python
text = """
Machine learning is an important branch of artificial
intelligence. It allows computers to learn patterns from
large amounts of data and make predictions without being
explicitly programmed for every situation.
"""


input_text = "summarize: " + text
```

```python
inputs = tokenizer(
    input_text,
    return_tensors="pt",
    max_length=128,
    truncation=True
)
```

```python
with torch.no_grad():

    output = model.generate(
        **inputs,
        max_length=50,
        num_beams=4,
        early_stopping=True
    )
```

```python
summary = tokenizer.decode(
    output[0],
    skip_special_tokens=True
)


print("\n===== SUMMARY =====")
print(summary)
```
