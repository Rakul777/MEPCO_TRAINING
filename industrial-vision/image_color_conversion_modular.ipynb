# Image Color Conversion — Modular IPython Notebook

This notebook performs BGR → Gray, BGR → RGB, and BGR → HSV conversion without using `cv2.cvtColor()` or `cv2.split()`.

## Block 1 — Import Libraries

```python
import cv2
import numpy as np
```

## Block 2 — Load Input Image

```python
image_path = r"C:\Users\Sudarsan\Desktop\iete2026.png"
img = cv2.imread(image_path)

if img is None:
    raise FileNotFoundError("Image not found. Check the image path.")

cv2.imshow("Original", img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Block 3 — BGR → GRAY (Scratch Implementation)

```python
height, width, _ = img.shape
gray = np.zeros((height, width), dtype=np.uint8)

for i in range(height):
    for j in range(width):
        B = img[i, j, 0]
        G = img[i, j, 1]
        R = img[i, j, 2]

        gray_value = 0.114 * B + 0.587 * G + 0.299 * R
        gray[i, j] = int(gray_value)

cv2.imshow("Gray", gray)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Block 4 — BGR → RGB (Scratch Implementation)

```python
rgb = np.zeros_like(img)

for i in range(height):
    for j in range(width):
        B = img[i, j, 0]
        G = img[i, j, 1]
        R = img[i, j, 2]

        rgb[i, j, 0] = R
        rgb[i, j, 1] = G
        rgb[i, j, 2] = B

cv2.imshow("RGB", rgb)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Block 5 — BGR → HSV (Scratch Implementation)

```python
hsv = np.zeros((height, width, 3), dtype=np.uint8)

for i in range(height):
    for j in range(width):
        B = img[i, j, 0] / 255.0
        G = img[i, j, 1] / 255.0
        R = img[i, j, 2] / 255.0

        maximum = max(R, G, B)
        minimum = min(R, G, B)
        diff = maximum - minimum

        if diff == 0:
            H = 0
        elif maximum == R:
            H = (60 * ((G - B) / diff)) % 360
        elif maximum == G:
            H = 60 * (((B - R) / diff) + 2)
        else:
            H = 60 * (((R - G) / diff) + 4)

        if maximum == 0:
            S = 0
        else:
            S = diff / maximum

        V = maximum

        hsv[i, j, 0] = int(H / 2)
        hsv[i, j, 1] = int(S * 255)
        hsv[i, j, 2] = int(V * 255)

cv2.imshow("HSV", hsv)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

## Block 6 — Display All Results

```python
cv2.imshow("Original", img)
cv2.imshow("Gray", gray)
cv2.imshow("RGB", rgb)
cv2.imshow("HSV", hsv)

cv2.waitKey(0)
cv2.destroyAllWindows()
```
