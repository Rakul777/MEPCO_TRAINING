```python
from tensorflow.keras.models import Model
from tensorflow.keras.layers import *
from tensorflow.keras.preprocessing.image import load_img
import numpy as np

# Load images
img1 = np.array(load_img("/content/image1.jpg", target_size=(128,128)))/255.0
img2 = np.array(load_img("/content/image2.jpg", target_size=(128,128)))/255.0

# Add batch dimension
img1 = np.expand_dims(img1, axis=0)
img2 = np.expand_dims(img2, axis=0)

# CNN
def cnn(i):
    x = Conv2D(32,(3,3),activation='relu')(i)
    x = MaxPooling2D((2,2))(x)
    x = Flatten()(x)
    x = Dense(64,activation='relu')(x)
    return x

# Inputs
i1 = Input((128,128,3))
i2 = Input((128,128,3))

# Features
f1 = cnn(i1)
f2 = cnn(i2)

# Squared difference
d = Subtract()([f1,f2])
d = Multiply()([d,d])

# Output
o = Dense(1,activation='sigmoid')(d)

# Model
model = Model([i1,i2],o)

# Compile
model.compile(optimizer='adam',loss='binary_crossentropy')

# Predict
p = model.predict([img1,img2])
print(p)
```
