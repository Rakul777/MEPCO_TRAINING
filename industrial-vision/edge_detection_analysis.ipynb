### Module 1: Load Image & Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt


# Load the image in grayscale mode
img = cv2.imread(r"C:\Users\rakul\OneDrive\Documents\Desktop\single-fresh-red-strawberry-on-table-green-background-food-fruit-sweet-macro-juicy-plant-image-photo.jpg", 0)
print(img)
plt.imshow(img, cmap='gray')
plt.axis('off')  # Hides coordinate axes
plt.show()
```

### Module 2: Standard Edge Detection (Canny & Sobel)

```python
# CANNY
c = cv2.Canny(img, 100, 200)
cv2.imshow("Canny", c)

# SOBEL
s = cv2.Sobel(img, cv2.CV_64F, 1, 1, ksize=3)
cv2.imshow("Sobel", s / 255)
```

### Module 3: Gaussian Noise Robustness Test

```python
# GAUSSIAN NOISE TEST
for i in [10, 30, 50]:
    n = np.random.normal(0, i, img.shape)
    g = np.uint8(np.clip(img + n, 0, 255))
    
    cv2.imshow("Gaussian" + str(i), g)
    cv2.imshow("Canny G" + str(i), cv2.Canny(g, 100, 200))
    cv2.imshow("Sobel G" + str(i), cv2.Sobel(g, cv2.CV_64F, 1, 1, 3) / 255)
```

### Module 4: Salt & Pepper Noise Robustness Test

```python
# SALT PEPPER TEST
for i in [0.01, 0.03, 0.05]:
    sp = img.copy()
    salt = np.random.rand(*img.shape) < i
    pepper = np.random.rand(*img.shape) < i
    
    sp[salt] = 255
    sp[pepper] = 0
    
    cv2.imshow("SP" + str(i), sp)
    cv2.imshow("Canny SP" + str(i), cv2.Canny(sp, 100, 200))
    cv2.imshow("Sobel SP" + str(i), cv2.Sobel(sp, cv2.CV_64F, 1, 1, 3) / 255)
```
