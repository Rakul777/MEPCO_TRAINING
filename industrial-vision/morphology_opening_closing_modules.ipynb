# Image Processing – Morphological Opening & Closing

## Module 1 – Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Module 2 – Load Grayscale Image

```python
img = cv2.imread(r"C:\Users\rakul\OneDrive\Documents\Desktop\cHJpdmF0ZS9sci9pbWFnZXMvd2Vic2l0ZS8yMDIyLTA0L2pvYjY3My0wNjktdi5qcGc.webp", 0)

print("Image loaded successfully")
print("Image shape:", img.shape)
```

## Module 3 – Display Original Image

```python
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

## Module 4 – Display Original Image Matrix

```python
print("Original Image Matrix:")
print(img)
```

## Module 5 – Rectangular Structural Element

```python
r = cv2.getStructuringElement(cv2.MORPH_RECT, (5, 5))

print("Rectangular Structural Element:")
print(r)
```

## Module 6 – Elliptical Structural Element

```python
e = cv2.getStructuringElement(cv2.MORPH_ELLIPSE, (5, 5))

print("Elliptical Structural Element:")
print(e)
```

## Module 7 – Cross Structural Element

```python
c = cv2.getStructuringElement(cv2.MORPH_CROSS, (5, 5))

print("Cross Structural Element:")
print(c)
```

## Module 8 – Opening with Rectangular Element

```python
op1 = cv2.morphologyEx(img, cv2.MORPH_OPEN, r)

plt.imshow(op1, cmap="gray")
plt.title("Rectangular Opening")
plt.axis("off")
plt.show()
```

## Module 9 – Rectangular Opening Matrix

```python
print("Rectangular Opening Matrix:")
print(op1)
```

## Module 10 – Opening with Elliptical Element

```python
op2 = cv2.morphologyEx(img, cv2.MORPH_OPEN, e)

plt.imshow(op2, cmap="gray")
plt.title("Elliptical Opening")
plt.axis("off")
plt.show()
```

## Module 11 – Elliptical Opening Matrix

```python
print("Elliptical Opening Matrix:")
print(op2)
```

## Module 12 – Opening with Cross Element

```python
op3 = cv2.morphologyEx(img, cv2.MORPH_OPEN, c)

plt.imshow(op3, cmap="gray")
plt.title("Cross Opening")
plt.axis("off")
plt.show()
```

## Module 13 – Cross Opening Matrix

```python
print("Cross Opening Matrix:")
print(op3)
```

## Module 14 – Closing with Rectangular Element

```python
cl1 = cv2.morphologyEx(img, cv2.MORPH_CLOSE, r)

plt.imshow(cl1, cmap="gray")
plt.title("Rectangular Closing")
plt.axis("off")
plt.show()
```

## Module 15 – Rectangular Closing Matrix

```python
print("Rectangular Closing Matrix:")
print(cl1)
```

## Module 16 – Closing with Elliptical Element

```python
cl2 = cv2.morphologyEx(img, cv2.MORPH_CLOSE, e)

plt.imshow(cl2, cmap="gray")
plt.title("Elliptical Closing")
plt.axis("off")
plt.show()
```

## Module 17 – Elliptical Closing Matrix

```python
print("Elliptical Closing Matrix:")
print(cl2)
```

## Module 18 – Closing with Cross Element

```python
cl3 = cv2.morphologyEx(img, cv2.MORPH_CLOSE, c)

plt.imshow(cl3, cmap="gray")
plt.title("Cross Closing")
plt.axis("off")
plt.show()
```

## Module 19 – Cross Closing Matrix

```python
print("Cross Closing Matrix:")
print(cl3)
```
