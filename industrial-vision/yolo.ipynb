```python
from ultralytics import YOLO
import cv2
from google.colab.patches import cv2_imshow

# Load YOLO model
model = YOLO("yolov8n.pt")

# Load image
img = cv2.imread("/content/bus.jpg")

# Detect objects
results = model(img)

# Show result
cv2_imshow(results[0].plot())

# Print detected objects
print(results[0].verbose())
```
