# Image Processing – Otsu, Adaptive Thresholding, Histogram 

## Module 1 – Import Libraries

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

## Module 2 – Load Grayscale Image

```python
img = cv2.imread(r"C:\Users\rakul\OneDrive\Documents\Desktop\images (1).jpg", 0)

print("Image loaded successfully")
print("Image shape:", img.shape)
```

## Module 3 – Display Original Image

```python
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")
plt.show()
```

## Module 4 – Display Original Image Matrix

```python
print("Original Image Matrix:")
print(img)
```

## Module 5 – Otsu Thresholding

```python
_, o = cv2.threshold(
    img, 0, 255,
    cv2.THRESH_BINARY + cv2.THRESH_OTSU
)

plt.imshow(o, cmap="gray")
plt.title("Otsu Thresholding")
plt.axis("off")
plt.show()
```

## Module 6 – Otsu Matrix

```python
print("Otsu Threshold Matrix:")
print(o)
```

## Module 7 – Adaptive Thresholding

```python
a = cv2.adaptiveThreshold(
    img, 255,
    cv2.ADAPTIVE_THRESH_MEAN_C,
    cv2.THRESH_BINARY,
    11, 2
)

plt.imshow(a, cmap="gray")200
plt.title("Adaptive Thresholding")
plt.axis("off")
plt.show()
```

## Module 8 – Adaptive Threshold Matrix

```python
print("Adaptive Threshold Matrix:")
print(a)
```

## Module 9 – Image Histogram

```python
plt.hist(img.ravel(), 256, [0, 256])
plt.title("Image Histogram")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```
